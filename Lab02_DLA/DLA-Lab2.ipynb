### Exercise 1.2: A Pre-trained BERT and Tokenizer

The model we will use is a *very* small BERT transformer called [DistilBERT](https://huggingface.co/distilbert/distilbert-base-uncased) this model was trained (using self-supervised learning) on the same corpus as BERT but using the full BERT base model as a *teacher*.

**Your next task**: Load the DistilBERT model and corresponding tokenizer. Use the tokenizer on a few samples from the dataset and pass the tokens through the model to see what outputs are provided. I suggest you use the [`AutoModel`](https://huggingface.co/transformers/v3.0.2/model_doc/auto.html) class (and the `from_pretrained()` method) to load the model and `AutoTokenizer` to load the tokenizer).

```python
# AutoClass imports.
from transformers import AutoTokenizer, AutoModel

# Your code here.
# Load the pretrained DistilBERT model (base encoder, no task-specific head yet)
# and its matching tokenizer.
model = AutoModel.from_pretrained('distilbert/distilbert-base-uncased')
tokenizer = AutoTokenizer.from_pretrained('distilbert/distilbert-base-uncased')
```

```python
# Encode a single sentence into token ids (as a PyTorch tensor), before feeding
# anything to the model.
tokens = tokenizer.encode('This is a sample sentence', return_tensors='pt')
```

```python
# Run a forward pass through DistilBERT. Since this is the base AutoModel
# (no classification head), the output is the raw transformer representation,
# not class logits.
output = model(tokens)
```

`output.last_hidden_state` contains the contextualized embedding of every input
token, at the output of the last transformer layer. Shape:
`(batch_size, sequence_length, hidden_size)` -- one 768-dimensional vector per
token. The first token (`last_hidden_state[0][0]`) corresponds to `[CLS]`,
whose embedding is conventionally used as a pooled, sentence-level representation.

```python
# Tokenize two examples together as a batch. Since the two sentences have
# different lengths, padding=True pads the shorter one with [PAD] tokens so
# both can be stacked into a single tensor.
batch = tokenizer(ds_train[:2]['text'], return_tensors='pt', padding=True)
```

```python
batch
```

```python
# Decode the padded batch to see the [PAD] tokens added to the shorter sequence.
tokenizer.decode(batch['input_ids'])
```

```python
# Compare running the model WITH and WITHOUT the attention mask:
# - no_masking: [PAD] tokens are treated as real input and can leak into the
#   self-attention computation for the other tokens.
# - masking: attention_mask tells the model which positions are real (1) and
#   which are padding (0), excluding padding from attention.
no_masking = model.forward(input_ids=batch['input_ids']).last_hidden_state
masking = model.forward(input_ids=batch['input_ids'], attention_mask=batch['attention_mask']).last_hidden_state
```

```python
# A non-zero difference here proves that padding DOES affect the
# representation of real tokens when the attention mask is not used --
# confirming why passing attention_mask is necessary with variable-length batches.
no_masking[0][0] - masking[0][0]
```

```python
masking = model.forward(**batch)
```

**Takeaway:** DistilBERT's raw output (`last_hidden_state`) gives one contextual
vector per token; the `[CLS]` token's vector is what we use as a summary
representation of the whole sentence. Padding tokens must always be masked out
via `attention_mask`, otherwise they silently distort the representation of
the real tokens, as shown above.

### Exercise 1.3: A Stable Baseline

In this exercise I want you to:
1. Use DistilBERT as a *feature extractor* to extract representations of the text strings from the dataset splits;
2. Train a classifier (your choice, by an SVM from Scikit-learn is an easy choice).
3. Evaluate performance on the validation and test splits.

These results are our *stable baseline* -- the **starting** point on which we will (hopefully) improve in the next exercise.

**Hint**: There are a number of ways to implement the feature extractor, but probably the best is to use a [feature extraction `pipeline`](https://huggingface.co/tasks/feature-extraction). You will need to interpret the output of the pipeline and extract only the `[CLS]` token from the *last* transformer layer. *How can you figure out which output that is?*

```python
from transformers import pipeline
import torch
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report

# Your code here.
# Use a feature-extraction pipeline to get DistilBERT's raw hidden states for
# every token of every training sentence, with no classification head involved.
extractor = pipeline('feature-extraction', model=model, tokenizer=tokenizer)
feats = extractor(list(ds_train['text']), return_tensors='pt')
```

**Understanding the pipeline output:** the cells below step through the nested
structure returned by the feature-extraction pipeline, to identify exactly
which tensor corresponds to the `[CLS]` embedding at the last layer --
`feats[0]` is the result for the first sentence, `feats[0][0]` is its
hidden-state tensor, and `feats[0][0][0]` is the first token's ([CLS]) vector,
shape `(768)`. This exploration justifies the indexing used in
`extract_cls_features` below.

```python
feats
```

```python
len(feats)
```

```python
feats[0]
```

```python
feats[0].shape
```

```python
feats[0][0]
```

```python
feats[0][0].shape
```

```python
feats[0][0][0]
```

```python
feats[0][0][0].shape
```

```python
# Stack the [CLS] vector of every sentence into a single (num_sentences, 768) tensor.
feats = torch.vstack([feat[0][0] for feat in feats])
```

```python
feats.shape
```

```python
# Wrap the logic explored above into a reusable function: run the
# feature-extraction pipeline on a list of texts and return only the [CLS]
# embedding (last layer) for each one, stacked into a single tensor.
def extract_cls_features(texts, extractor):
    feats = extractor(list(texts), return_tensors='pt')
    return torch.vstack([feat[0][0] for feat in feats])
```

```python
# Extract frozen DistilBERT [CLS] features for all three splits. No fine-tuning
# happens here -- this is what makes the classifier below a "stable baseline":
# the representation is fixed, only the downstream classifier is trained.
extractor = pipeline('feature-extraction', model=model, tokenizer=tokenizer)

X_train = extract_cls_features(ds_dict['train']['text'], extractor)
X_val   = extract_cls_features(ds_dict['validation']['text'], extractor)
X_test  = extract_cls_features(ds_dict['test']['text'], extractor)

y_train = ds_dict['train']['label']
y_val   = ds_dict['validation']['label']
y_test  = ds_dict['test']['label']
```

```python
# Train a linear SVM on top of the frozen CLS features, using scikit-learn's
# default regularization strength (C=1.0).
clf = LinearSVC()
clf.fit(X_train, y_train)
```

```python
# Evaluate the baseline on both validation and test splits.
print("Validation:")
print(classification_report(y_val, clf.predict(X_val)))

print("Test:")
print(classification_report(y_test, clf.predict(X_test)))
```

**Baseline result:** with the default C, the SVM reaches ~0.82 accuracy on
validation and ~0.80 on test. Since C controls the strength of L2
regularization and its effect is multiplicative rather than additive, it makes
sense to search over a logarithmic range of values rather than trying a few
values linearly, to check whether the default is actually a good choice.

```python
# Search for the best regularization strength C over a logarithmic grid
# (from 1e-4 to 1e4), selecting the value that maximizes accuracy on the
# validation split.
C_values = np.logspace(-4, 4, 20)  # from 0.0001 to 10000
best_C, best_acc = None, 0

for C in C_values:
    clf_tmp = LinearSVC(C=C)
    clf_tmp.fit(X_train, y_train)
    acc = clf_tmp.score(X_val, y_val)
    if acc > best_acc:
        best_acc = acc
        best_C = C

print(f"Best C: {best_C:.4f}, Val Acc: {best_acc:.4f}")
```

```python
# Refit the SVM with the best C found above, then re-evaluate on validation and test.
clf = LinearSVC(C=best_C)
clf.fit(X_train, y_train)

print("Validation:")
print(classification_report(y_val, clf.predict(X_val)))

print("Test:")
print(classification_report(y_test, clf.predict(X_test)))
```

**Considerations on the baseline:** tuning C (best value found: ~0.089) does
not meaningfully change performance -- accuracy stays at 0.82 on validation
and 0.80 on test, with only third/fourth-decimal differences and no consistent
improvement across classes. This confirms the bottleneck of this baseline is
not the linear classifier's regularization, but the quality of the frozen
DistilBERT features themselves: since they come from a model *not trained for
this task*, a wide range of C values already separate the classes about as
well as they can. This motivates fine-tuning DistilBERT directly in Exercise 2,
so the representation itself can adapt to the sentiment classification task
instead of staying fixed.

### Exercise 2.1: Token Preprocessing

The first thing we need to do is *tokenize* our dataset splits -- we don't want to re-tokenize our inputs for every batch! Our current datasets return a dictionary with *strings*, but we want *input token ids* (i.e. the output of the tokenizer). This is easy enough to do by hand, but the Hugging Face `Dataset` class provides convenient, efficient, and *lazy* methods. See the documentation for [`Dataset.map`](https://huggingface.co/docs/datasets/v3.5.0/en/package_reference/main_classes#datasets.Dataset.map).

**Tip**: Verify that your new datasets are returning for every element: `text`, `label`, `intput_ids`, and `attention_mask`.

```python
# Your code here.
# Tokenize every example (truncating sequences that exceed the model's max
# length) so training doesn't need to re-tokenize text on every batch.
def preprocess_function(examples):
    return tokenizer(examples['text'], truncation=True)

# Dataset maps are NOT performed in-place!
tokenized_train = ds_train.map(preprocess_function, batched=True)
tokenized_val = ds_dict['validation'].map(preprocess_function, batched=True)
tokenized_test = ds_dict['test'].map(preprocess_function, batched=True)
```

fields (compare with the un-tokenized `ds_train`). `set_format('pt', ...)`

Trainer/DataCollator downstream.

### Exercise 2.3: Fine-tuning DistilBERT

Finally. In this exercise you should use a HuggingFace [`Trainer`](https://huggingface.co/docs/transformers/main/en/trainer) to fine-tune your model on the Rotten Tomatoes training split. Setting up the trainer will involve (at least):


1. Instantiating a [`DataCollatorWithPadding`](https://huggingface.co/docs/transformers/en/main_classes/data_collator) object which is what *actually* does your batch construction (by padding all sequences to the same length).
2. Writing an *evaluation function* that will measure the classification accuracy. This function takes a single argument which is a tuple containing `(logits, labels)` which you should use to compute classification accuracy (and maybe other metrics like F1 score, precision, recall) and return a `dict` with these metrics.  
3. Instantiating a [`TrainingArguments`](https://huggingface.co/docs/transformers/v4.51.1/en/main_classes/trainer#transformers.TrainingArguments) object using some reasonable defaults.
4. Instantiating a `Trainer` object using your train and validation splits, you data collator, and function to compute performance metrics.
5. Calling `trainer.train()`, waiting, waiting some more, and then calling `trainer.evaluate()` to see how it did.

**Tip**: When prototyping this laboratory I discovered the HuggingFace [Evaluate library](https://huggingface.co/docs/evaluate/en/index) which provides evaluation metrics. However I found it to have insufferable layers of abstraction and getting actual metrics computed. I suggest just using the Scikit-learn metrics...

```python
# Your code here.
from transformers import TrainingArguments, Trainer, DataCollatorWithPadding
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

# Metric function required by the Trainer: receives (logits, labels) at each
# evaluation step and returns a dict of scalar metrics to log/track.
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    precision, recall, f1, _ = precision_recall_fscore_support(labels, preds, average='binary')
    acc = accuracy_score(labels, preds)
    return {'accuracy': acc, 'precision': precision, 'recall': recall, 'f1': f1}

# Dynamically pads each batch to the length of its longest sequence (rather
# than the fixed max_length used above for exploration) -- more efficient.
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
```

```python
def finetune_and_test(training_args):
    """
    Fine-tune a FRESH DistilBertForSequenceClassification (re-instantiated from
    the pretrained checkpoint) using the given TrainingArguments, then return
    the trainer and its predictions on the held-out test split.

    Re-instantiating the model on every call is essential: reusing the same
    model object across calls would mean each new configuration continues
    training from the previous run's already fine-tuned weights, making
    configurations impossible to compare fairly.
    """
    model = DistilBertForSequenceClassification.from_pretrained(
        'distilbert/distilbert-base-uncased', num_labels=2
    )
    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_train,
        eval_dataset=tokenized_val,
        data_collator=data_collator,
        compute_metrics=compute_metrics,
    )
    trainer.train()
    # load_best_model_at_end=True (set in every TrainingArguments below) means
    # `trainer` now holds the checkpoint with the best validation F1, not
    # necessarily the one from the last epoch.
    print(f"Best checkpoint: {trainer.state.best_model_checkpoint}, best F1 (val): {trainer.state.best_metric:.4f}")
    return trainer, trainer.predict(tokenized_test)
```

```python
# Configuration 1/3 -- "baseline": the straightforward choice suggested by the
# exercise (lr=2e-5, 10 epochs). load_best_model_at_end ensures the final test
# evaluation uses the checkpoint with the best validation F1 across all 10
# epochs, not just whichever epoch happens to be last.
training_args_baseline = TrainingArguments(
    output_dir='./output/baseline',
    learning_rate=2e-5,
    per_device_train_batch_size=48,
    per_device_eval_batch_size=48,
    num_train_epochs=10,
    use_cpu=False,
    save_strategy='epoch',
    report_to='wandb',
    run_name='baseline_lr2e-5_ep10',
    logging_strategy="steps",
    logging_steps=1,
    do_eval=True,
    eval_strategy='epoch',
    load_best_model_at_end=True,
    metric_for_best_model='f1',
    greater_is_better=True,
    save_total_limit=2,   # keep only the best checkpoint + the last one
)
trainer_baseline, preds_baseline = finetune_and_test(training_args_baseline)
print("Baseline (lr=2e-5, 10 epochs):")
print(classification_report(tokenized_test['label'], np.argmax(preds_baseline.predictions, axis=-1)))
```

```python

# Configuration 2/3 -- same learning rate as the baseline, but only 3 epochs.
# Isolates the effect of training duration, keeping every other hyperparameter
# identical to the baseline.
training_args_3ep = TrainingArguments(
    output_dir='./output/epochs3',
    learning_rate=2e-5,
    per_device_train_batch_size=48,
    per_device_eval_batch_size=48,
    num_train_epochs=3,
    use_cpu=False,
    save_strategy='epoch',
    report_to='wandb',
    run_name='lr2e-5_ep3',
    logging_strategy="steps",
    logging_steps=1,
    do_eval=True,
    eval_strategy='epoch',
    load_best_model_at_end=True,
    metric_for_best_model='f1',
    greater_is_better=True,
    save_total_limit=2,   # keep only the best checkpoint + the last one
)
trainer_3ep, preds_3ep = finetune_and_test(training_args_3ep)
print("Config epochs=3 (lr=2e-5):")
print(classification_report(tokenized_test['label'], np.argmax(preds_3ep.predictions, axis=-1)))
```

```python
# Configuration 3/3 -- same 10 epochs as the baseline, but a higher learning
# rate (5e-5, within the range BERT's original paper recommends). Isolates the
# effect of learning rate, changing only one variable relative to the baseline.
training_args_lr5 = TrainingArguments(
    output_dir='./output/lr5e-5',
    learning_rate=5e-5,
    per_device_train_batch_size=48,
    per_device_eval_batch_size=48,
    num_train_epochs=10,
    use_cpu=False,
    save_strategy='epoch',
    report_to='wandb',
    run_name='lr5e-5_ep10',
    logging_strategy="steps",
    logging_steps=1,
    do_eval=True,
    eval_strategy='epoch',
    load_best_model_at_end=True,
    metric_for_best_model='f1',
    greater_is_better=True,
    save_total_limit=2,   # keep only the best checkpoint + the last one
)
trainer_lr5, preds_lr5 = finetune_and_test(training_args_lr5)
print("Config lr=5e-5 (10 epochs):")
print(classification_report(tokenized_test['label'], np.argmax(preds_lr5.predictions, axis=-1)))
```

```python
# Compare all three configurations side by side on the test split, using each
# configuration's best validation checkpoint (not necessarily the last epoch).
import pandas as pd

results = {
    'baseline (lr=2e-5, ep=10)': classification_report(tokenized_test['label'], np.argmax(preds_baseline.predictions, axis=-1), output_dict=True),
    'ep=3 (lr=2e-5)': classification_report(tokenized_test['label'], np.argmax(preds_3ep.predictions, axis=-1), output_dict=True),
    'lr=5e-5 (ep=10)': classification_report(tokenized_test['label'], np.argmax(preds_lr5.predictions, axis=-1), output_dict=True),
}

summary = pd.DataFrame({
    name: {'accuracy': r['accuracy'], 'f1 (class 1)': r['1']['f1-score']}
    for name, r in results.items()
}).T
summary
```

## Considerations

All three configurations show substantial overfitting on the training loss
(collapsing to values as low as 0.0001-0.003 within a few epochs). Validation
accuracy, however, stabilizes quickly (within the first 2-4 epochs) and stays
in a narrow ~0.83-0.85 band regardless of how much longer training continues
-- except with the higher learning rate (5e-5), where more aggressive
overfitting starts to visibly hurt validation accuracy too, not just the
training loss (its best checkpoint occurs at epoch 1).

After correcting for this with `load_best_model_at_end`, the epochs=3 and
lr=5e-5 configurations reach an almost identical F1 score on the positive
class (~0.847) via different paths (stopping early vs. converging faster),
while the standard 10-epoch/lr=2e-5 configuration is the weakest of the three
even at its own best checkpoint (epoch 6). This suggests that with this
pretrained backbone, useful learning happens within the first few epochs, and
training longer with a conservative learning rate does not make up for it.

With only one run per configuration on a ~1,000-example test split, these
differences are close to the expected sampling noise -- this experiment is
more informative about the *qualitative* overfitting behaviour of the model
than about definitively ranking the three configurations.
